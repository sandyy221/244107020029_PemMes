```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('/content/drive/MyDrive/PML_2026/Titanic-Dataset.csv')
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
# Age - mean
df['Age'].fillna(value=df['Age'].mean(), inplace=True)

# Cabin - "DECK"
df['Cabin'].fillna(value="DECK", inplace=True)

# Embarked - modus
df['Embarked'].fillna(value=df['Embarked'].mode, inplace=True)
```

```python
df.info()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# === EKSTRAKSI FITUR ===

# 1. FamilySize: jumlah keluarga di kapal (SibSp + Parch + diri sendiri)
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1

# 2. Title: ambil gelar dari kolom Name (Mr, Mrs, Miss, Master, dll.)
df["Title"] = df["Name"].str.extract(r',\s*([^\.]+)\.', expand=False)
title_mapping = {
    "Mlle": "Miss", "Ms": "Miss", "Mme": "Mrs",
    "Lady": "Rare", "Countess": "Rare", "Capt": "Rare", "Col": "Rare",
    "Don": "Rare", "Dr": "Rare", "Major": "Rare", "Rev": "Rare",
    "Sir": "Rare", "Jonkheer": "Rare", "Dona": "Rare"
}
df["Title"] = df["Title"].replace(title_mapping)

# 3. AgeBin: kelompokkan usia menjadi kategori
df["AgeBin"] = pd.cut(df["Age"], bins=[0, 12, 18, 35, 60, 100],
                      labels=["Child", "Teen", "YoungAdult", "Adult", "Senior"])

# 4. CabinDeck: ambil huruf pertama dari kolom Cabin
df["CabinDeck"] = df["Cabin"].astype(str).str[0]

# 5. Fare per person: bagi harga tiket dengan jumlah keluarga
df["FarePerPerson"] = df["Fare"] / df["FamilySize"]

print("✅ Hasil Ekstraksi fitur ")
df[["Name", "Title", "Age", "AgeBin", "FamilySize", "Cabin", "CabinDeck", "FarePerPerson"]].head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# === VISUALISASI FITUR HASIL EKSTRAKSI ===

plt.figure(figsize=(12, 6))
sns.barplot(x="FamilySize", y="Survived", data=df)
plt.title("Survival Rate berdasarkan Family Size")
plt.show()

plt.figure(figsize=(12, 6))
sns.barplot(x="Title", y="Survived", data=df,
            order=df.groupby("Title")["Survived"].mean().sort_values().index)
plt.title("Survival Rate berdasarkan Title")
plt.xticks(rotation=45)
plt.show()

plt.figure(figsize=(12, 6))
sns.barplot(x="AgeBin", y="Survived", data=df,
            order=["Child", "Teen", "YoungAdult", "Adult", "Senior"])
plt.title("Survival Rate berdasarkan Kelompok Usia (AgeBin)")
plt.show()

plt.figure(figsize=(12, 6))
sns.barplot(x="CabinDeck", y="Survived", data=df,
            order=df.groupby("CabinDeck")["Survived"].mean().sort_values().index)
plt.title("Survival Rate berdasarkan Cabin Deck")
plt.show()
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler

# 1. Memuat dataset wbc.csv
df = pd.read_csv('/content/drive/MyDrive/PML_2026/wbc.csv')

# Menghapus kolom 'id' (tidak prediktif) dan 'Unnamed: 32'
kolom_tidak_digunakan = ['id', 'Unnamed: 32']
df_clean = df.drop(columns=[col for col in kolom_tidak_digunakan if col in df.columns])

# Memisahkan Fitur (X) dan Target (y)
X = df_clean.drop(columns=['diagnosis']) # 30 fitur numerik yang dapat digunakan
y = df_clean['diagnosis']                # Variabel target
```

```python
# STEP 2: Process Encoding pada kolom "diagnosis"
le = LabelEncoder()
y_encoded = le.fit_transform(y)

# Menampilkan pemetaan hasil encoding
print("Pemetaan Label Encoding 'diagnosis':")
for class_label, index in zip(le.classes_, le.transform(le.classes_)):
    print(f"  {class_label} -> {index}")
```

```python
#STEP 3: Standardisasi pada semua kolom numerik
scaler = StandardScaler()

# Melakukan scaling (mean = 0, std = 1) pada seluruh fitur numerik X
X_scaled_array = scaler.fit_transform(X)

# Mengubah kembali array NumPy hasil scaling menjadi DataFrame Pandas
X_scaled = pd.DataFrame(X_scaled_array, columns=X.columns)

print("\nRINGKASAN PEMBEDAAN VARIABEL")
print(f"Total kolom awal          : {df.shape[1]}")
print(f"Kolom dibuang             : {kolom_tidak_digunakan}")
print(f"Total fitur yang digunakan: {X_scaled.shape[1]} kolom numerik")
print(f"Jumlah sampel/baris       : {X_scaled.shape[0]}")
X_scaled.head()
```
